# Indexação e Fatiamento de Arrays

```python
import numpy as np
```

```python
array = np.arange(0, 30, 3)
```

```python
array
```

```python
array[3]
```

```python
array[0:2]
```

```python
array[:5]
```

```python
array[2:]
```

```python
array[[0, 5]]
```

```python
array[4:7]
```

```python
array[5:] = 100
```

```python
array
```

```python
#.arange faz o array e .reshape faz linha x coluna
array_2d = np.arange(50).reshape([5,10])
```

```python
array_2d
```

```python
array_2d[0:3]
```

```python
array_2d[0:3, 0:3]
```

```python
array_2d[0:3, 4:]
```

```python
array_2d > 20
```

```python
booleano = array_2d > 20
```

```python
array_2d[booleano]
```

```python
np.logical_and(array_2d >= 20, array_2d <= 30)
```

```python
booleano = np.logical_and(array_2d >= 20, array_2d <= 30)
```

```python
array_2d[booleano]
```

```python
np.logical_or(array_2d >= 20, array_2d <= 10)
```
