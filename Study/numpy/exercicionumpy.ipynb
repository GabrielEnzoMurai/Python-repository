```python
import numpy as np
```

```python
# Criando um array de 1 a 10
array1 = np.arange(1, 11)
print("Array de 10:", array1)
```

```python
# Criando um array de 0
array2 = np.zeros(10)
print("Array de zeros:", array2)
```

```python
# criando um linspace de 10 valores entre 0 e 50
array_linspace = np.linspace(0, 50, 10)
print("Array linspace de 0 a 50 com 10 valores:", array_linspace)
```

```python
# Criando uma matriz identidade 3x3
matriz_identidade = np.eye(3)
print("Matriz identidade 3x3:\n", matriz_identidade)
```

```python
# Indexação e fatiamento
arr = np.arange(10, 70, 10)
fatiando = arr[:3]
fatiando
```

```python
# Calculando o quadrado dos números de 1 a 5
arr_quadrado = np.arange(1, 6) ** 2
arr_quadrado
```

```python
# Soma de arrays
arr_a = np.arange(1, 4)
arr_b = np.arange(4, 7)

array_soma = arr_a + arr_b
array_soma
```

```python
# Transformando um array de 10 em matriz 3x3
arr_reshape = np.arange(1, 10).reshape(3, 3)
arr_reshape
```

```python
# Criando um array 4x4 com valores aleatórios entre 0 e 1
arr_random = np.random.rand(4, 4)
arr_random
```
