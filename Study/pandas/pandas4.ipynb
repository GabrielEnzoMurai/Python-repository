```python
def valor_agregado(valor):
    if valor >= 1000:
        return "Alto valor"
    else:
        return "Baixo valor"
```

```python
valor_agregado(1000)
```

```python
df["Valor"].apply(valor_agregado)
```

```python
df["Valor Agregado"] = df["Valor"].apply(valor_agregado)
```

```python
df
```

```python
def substitui(nome_produto):
    if nome_produto == "Notebook":
        return "Computador"
    else:
        return nome_produto
```

```python
df["Produto"].apply(substitui)
```

```python
df["Produto"].apply(lambda nome_produto: "Computador" if nome_produto == "Notebook" else nome_produto)
```

```python
df["coluna_teste"] = df["Valor"].apply(lambda valor: "Mais de 500" if valor > 500 else "Menos de 500")
df
```
