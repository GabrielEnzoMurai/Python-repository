```python
import numpy as np
import pandas as pd
```

```python
dados = [
    {"Nome": "Ana", "Departamento": "TI", "Salário": 6000, "Estado": "São Paulo"},
    {"Nome": "Bruno", "Departamento": "TI", "Salário": 6000, "Estado": "Rio de Janeiro"},
    {"Nome": "Carla", "Departamento": "RH", "Salário": 4000, "Estado": "São Paulo"},
    {"Nome": "Daniel", "Departamento": "RH", "Salário": 4200, "Estado": "Belo Horizonte"},
    {"Nome": "Eduardo", "Departamento": "Vendas", "Salário": 3000, "Estado": "São Paulo"},
    {"Nome": "Fernanda", "Departamento": "Vendas", "Salário": 3500, "Estado": "Rio de Janeiro"},
]
```

```python
df = pd.DataFrame(dados)
df
```

```python
df.groupby("Departamento")["Salário"].mean()
```

```python
df.groupby("Departamento")["Salário"].max()
```

```python
df.groupby("Departamento")["Salário"].min()
```

```python
df.groupby("Departamento")["Salário"].count()
```

```python
df.groupby("Departamento")["Salário"].count()
```

```python
df.groupby("Departamento")["Salário"].nunique()
```

```python
df.groupby("Estado")["Salário"].sum()
```
