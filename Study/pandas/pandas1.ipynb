```python
import pandas as pd
import numpy as np
```

```python
np.nan
```

```python
dados = {"A": [1, 2, np.nan], "B": [5, np.nan, np.nan], "C": [1, 2, 3]}
```

```python
dados
```

```python
df = pd.DataFrame(dados)
df
```

```python
df.dropna()
```

```python
# Dropa apenas as linhas que possuem 2+ dados faltantes
df.dropna(thresh=2)
```

```python
# axis = 0 linha
# axis = 1 coluna
df.dropna(axis=1, thresh=2)
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.fillna(value="Tá faltando")
```

```python
df["A"].fillna(value="Tá faltando")
```

```python
df["A"] = df["A"].fillna(value="Tá faltando")
df
```

```python
df.fillna({"B": "Tá faltando também"})
```

```python
df.fillna({"A": df["A"].mean(), "B": "Tá faltando"})
```
