```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 100, 30)
```

```python
x
```

```python
y = x ** 2
```

```python
y
```

```python
plt.plot(x, y)
```

```python
plt.plot(x, y)
plt.xlabel("Valores Eixo X")
```

```python
plt.plot(x, y)
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
```

```python
plt.plot(x, y)
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```

```python
plt.plot(x, y, color="red")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```

```python
plt.plot(x, y, "r--")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```

```python
plt.bar(x, y)
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```

```python
plt.subplot(1, 2, 1)
```

```python
plt.subplot(1, 2, 1)
plt.subplot(1, 2, 2)
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y, "r--")
plt.subplot(1, 2, 2)
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y, "r--")
plt.subplot(1, 2, 2)
plt.plot(x, y ** 2, "g*-")
```

```python
plt.subplot(2, 2, 1)
plt.plot(x, y, "g--")
plt.subplot(2, 2, 2)
plt.subplot(2, 2, 3)
plt.subplot(2, 2, 4)
plt.plot(x ** 2, y, "y--")
```

```python
plt.figure()
```

```python
# Primeiro argumento é horizontal e o segundo vertical

plt.figure(figsize=(15, 10))
plt.subplot(1, 2, 1)
plt.plot(x, y, "r--")
plt.subplot(1, 2, 2)
plt.plot(x, y, "b--")
```
