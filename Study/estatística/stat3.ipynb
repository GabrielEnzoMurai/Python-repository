### Imports

```python
import pandas as pd
import numpy as np
```

### Lendo Dados

```python
df = pd.read_csv("ifood-restaurants-february-2021.csv", sep=",", header=0)
```

```python
df.head(5)
```

```python
df.info()
```

### Selecionando váriaveis relevantes

```python
df_sel = df[[
    "availableForScheduling",
    "category",
    "delivery_fee",
    "delivery_time",
    "price_range",
    "minimumOrderValue"
]]
```

```python
num = df_sel.select_dtypes(include=['number']).columns.tolist()
cat = df_sel.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()

print("Variáveis Numéricas:", num)
print("Variáveis Categóricas:", cat)
```

```python
df_sel.describe()
```

```python
df_sel["category"].mode()
```

```python
df_sel["category"].mode()[0]
```

```python
df_sel["availableForScheduling"].mode()[0]
```

```python
df_sel["price_range"].mode()[0]
```

```python
for var in cat:
    count = df_sel[var].value_counts()

    if len(count) >= 5:
        print("--------------------------")
        print(f"Coluna: {var}")
        print(f"Moda: {df_sel[var].mode()[0]}\n")
        print(f"{count.iloc[0:5]}\n")
    else:
        print("--------------------------")
        print(f"Coluna: {var}")
        print(f"Moda: {df_sel[var].mode()[0]}\n")
        print(f"{count.iloc[0:5]}\n")
```
