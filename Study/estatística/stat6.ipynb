# Boxplots

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("ifood-restaurants-february-2021.csv", sep=",", header=0)
```

```python
df
```

```python
num = df.select_dtypes(include=['number']).columns.to_list()
num
```

```python
plt.figure(figsize=(8, 6))
plt.boxplot(df['delivery_time'], labels=['Tempo de Entrega'])
plt.title('Boxplot Tempo de Entrega')
plt.ylabel('Valores')
plt.show()
```

```python
# Tirar os outliers calculando o Interquartil

Q1 = df['delivery_time'].quantile(0.25)
Q3 = df['delivery_time'].quantile(0.75)
IQR = Q3 - Q1
```

```python
df_out = df[((df['delivery_time'] < (Q1 - 1.5 * IQR)) | (df['delivery_time'] > (Q3 + 1.5 * IQR)))]
```

```python
df_out.head(3)
```

```python
print("Quantidade de registro na base:", len(df))
```

```python
print(f"Quantidade de Outliers: {len(df_out)}")
print(f"Percentual de Outliers: {np.round(100 * len(df_out) / len(df), 2)}%")
```

```python
df_sem_out = df[~((df['delivery_time'] < (Q1 - 1.5 * IQR)) | (df['delivery_time'] > (Q3 + 1.5 * IQR)))]
len(df_sem_out)
```

```python
plt.figure(figsize=(8, 6))
plt.boxplot(df_sem_out['delivery_time'], labels=['Tempo de Entrega'])
plt.title('Boxplot Tempo de Entrega')
plt.ylabel('Valores')
plt.show()
```
