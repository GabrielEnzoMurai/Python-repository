### Exercício 2 - Média Ponderada

```python
notas = np.array([8.5, 7.0])
pesos = np.array([0.6, 0.4])

media_ponderada = np.average(notas, weights=pesos)

media_ponderada
```

```python
# Se o peso de 0.4 fosse 0.5]

notas = np.array([8.5, 7.0])
pesos = np.array([0.5, 0.5])

media_ponderada = np.average(notas, weights=pesos)

media_ponderada
```
