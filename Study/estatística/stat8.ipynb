# Distribuições

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
n_simulacoes = 1000000

somas = []

for rodada in range(n_simulacoes):
    dado1 = np.random.randint(1, 7, 1)
    dado2 = np.random.randint(1, 7, 1)

    soma = dado1 + dado2
    somas.append(int(soma[0]))

plt.hist(somas, bins=11)
plt.title("Distribuições da Soma de Dois Dados de 6 Faces")
plt.xlabel("Soma dos Resultados")
plt.ylabel("Frequência Relativa")
plt.xticks(range(2, 13))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
