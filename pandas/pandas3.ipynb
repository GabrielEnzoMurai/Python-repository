```python
import numpy as np
import pandas as pd
```

```python
usuarios = {
    "ID_usuario": [1, 2, 3],
    "Nome": ["Maria", "Bruno", "Joao"],
    "Cidade": ["São Paulo", "Rio", "Belo Horizonte"]
}

compras = {
    "ID_usuario": [1, 2, 1, 1, 3, 3, 2, 3],
    "Produto": ["Notebook", "Camera", "Camiseta", "Camiseta", "Caneca", "Caneca", "Calça", "Calça"],
    "Valor": [2000, 1000, 70, 60, 50, 50, 120, 100]
}

usuarios_novos = {
    "ID_usuario": [4, 5],
    "Nome": ["Pedro", "Marcos"]
}
```

```python
usuarios_df = pd.DataFrame(usuarios)
compras_df = pd.DataFrame(compras)
usuarios_novos_df = pd.DataFrame(usuarios_novos)
```

```python
pd.concat([usuarios_df, usuarios_novos_df])
```

```python
usuarios_geral = pd.concat([usuarios_df, usuarios_novos_df])
```

```python
usuarios_geral
```

```python
usuarios_geral.fillna({"Cidade": "Cidade Desconhecida"}, inplace=True)
```

```python
usuarios_geral
```

```python
compras_df
```

```python
# Left => manter valores do df da esquerda e associar aos valores da direita
# Right => manter valores do df da direita e associar aos valores da esquerda
# Inner => apenas o que tem em comum

compras_df.merge(usuarios_geral, how="left", on="ID_usuario")
```

```python
compras_df.merge(usuarios_geral, how="right", on="ID_usuario")
```

```python
compras_df.merge(usuarios_geral, how="inner", on="ID_usuario")
```
