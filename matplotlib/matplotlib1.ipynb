```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 100, 30)
```

```python
x
```

```python
y = x ** 2
```

```python
y
```

```python
plt.plot(x, y)
```

```python
plt.plot(x, y)
plt.xlabel("Valores Eixo X")
```

```python
plt.plot(x, y)
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
```

```python
plt.plot(x, y)
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```

```python
plt.plot(x, y, color="red")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```

```python
plt.plot(x, y, "r--")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```

```python
plt.bar(x, y)
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Meu gráfico de Exemplo")
```
