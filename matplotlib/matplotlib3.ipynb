```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Salaries.csv")
df
```

```python
df.head()
```

```python
df = df[df["BasePay"] != "Not Provided"]
df
```

```python
df["BasePay"] = df["BasePay"].astype(float)
```

```python
df_plots = df[["BasePay", "Year", "JobTitle"]]
```

```python
df_plots
```

```python
# Parecido com .count mas pega o tamanho da base inteira

df_plots["Year"] = df["Year"].astype(str)
qtd_por_ano = df_plots.groupby("Year").size()
qtd_por_ano
```

```python
plt.figure(figsize=(8, 5))

plt.bar(qtd_por_ano.index, qtd_por_ano.values)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(qtd_por_ano.index, qtd_por_ano.values, color="g")
plt.title("Quantidade por Ano")
plt.xlabel("Ano")
plt.ylabel("Valor")
```

```python

media_salario = df.groupby("Year")["BasePay"].mean()
media_salario
```

```python
plt.figure(figsize=(8, 5))

plt.plot(media_salario.index, media_salario.values)
plt.title("Média Salarial por Ano")
plt.xlabel("Ano")
plt.ylabel("Média")
```

```python
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(10, 5))

ax[0].bar(qtd_por_ano.index, qtd_por_ano.values, color="g")
ax[0].set_title("Quantidade por Ano")
ax[0].set_xlabel("Ano")
ax[0].set_ylabel("Quantidade")

ax[1].plot(media_salario.index, media_salario.values)
ax[1].set_title("Média Salarial por Ano")
ax[1].set_xlabel("Ano")
ax[1].set_ylabel("Média")
```
