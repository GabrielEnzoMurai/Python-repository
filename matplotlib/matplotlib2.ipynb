```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 100, 30)
y = x ** 2
```

```python
fig, ax = plt.subplots(2, 2)
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10, 5))

ax[0, 0].plot(x, y, "r--")
ax[0, 0].set_xlabel("Eixo x primeiro graf")

ax[0, 1].plot(x, y, "g-.")

ax[1, 0].plot(x, y**2, "b*-")

ax[1, 1].bar(x, y)

# Ajusta os gráficos
plt.tight_layout()
```
